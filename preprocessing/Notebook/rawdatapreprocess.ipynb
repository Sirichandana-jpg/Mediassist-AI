```python
import pandas as pd

df = pd.read_csv(r"C:\PROJECT\data\dataset.csv")

df.head()
```

```python
df
```

```python
symptom_cols = df.columns[1:]

df["symptom_list"] = df[symptom_cols].apply(
    lambda x: [s for s in x if pd.notna(s)], axis=1
)
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
cols = df.columns.tolist()

# Keep last column as it is
new_cols = ["Disease"] + [f"Symptom_{i}" for i in range(1, len(cols)-1)] + [cols[-1]]

df.columns = new_cols
df.head()
```

```python
df
```

```python
df.columns = df.columns.str.strip() #to clean column names(hidden spaces)
```

```python
print(df.columns)
```

```python
df["symptom_tuple"] = df["symptom_list"].apply(tuple) #convert list to tuple as list is unhashable

df.drop_duplicates(subset=["Disease", "symptom_tuple"], inplace=True)
```

```python
df.head()
```

```python
df
```

```python
df.drop(columns=["symptom_tuple"], inplace=True)
df.head()
```

```python
df["symptoms_text"] = df["symptom_list"].apply(lambda x: " ".join(x))
df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["symptoms_text"])
```

```python
df.head()
```

```python
sev_df = pd.read_csv(r"C:\PROJECT\BIG DATA\Symptom_severity.csv",header=None)
```

```python
print(sev_df.head())
print(sev_df.columns)
```

```python
sev_df.columns = ["Symptom", "weight"]
```

```python
print(sev_df.head())
```

```python
sev_df["Symptom"] = sev_df["Symptom"].str.strip().str.lower()
sev_df.head()
```

```python
severity_dict = dict(zip(
    sev_df["Symptom"],
    sev_df["weight"]
))
```

```python
severity_dict
```

```python
def calculate_severity(symptoms):
    return sum(severity_dict.get(sym, 0) for sym in symptoms)

df["severity_score"] = df["symptom_list"].apply(calculate_severity)
```

```python
import numpy as np
from scipy.sparse import hstack

severity_feature = np.array(df["severity_score"]).reshape(-1,1)

X = hstack([X, severity_feature])
```

```python
import matplotlib.pyplot as plt

df["Disease"].value_counts().plot(kind="bar")
plt.title("Disease Distribution")
plt.xlabel("Disease")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
df["severity_score"].plot(kind="hist", bins=20)
plt.title("Severity Score Distribution")
plt.xlabel("Severity Score")
plt.ylabel("Frequency")
plt.show()
```

```python
df["num_symptoms"] = df["symptom_list"].apply(len)

df["num_symptoms"].plot(kind="hist", bins=10)
plt.title("Number of Symptoms Distribution")
plt.xlabel("Number of Symptoms")
plt.ylabel("Frequency")
plt.show()
```

```python
from collections import Counter

all_symptoms = [sym for row in df["symptom_list"] for sym in row]

symptom_counts = Counter(all_symptoms)

common = symptom_counts.most_common(20)

symptoms, counts = zip(*common)

plt.bar(symptoms, counts)
plt.xticks(rotation=90)
plt.title("Top 20 Symptoms")
plt.show()
```
