```python
df["status"].value_counts().plot(
    kind="bar"
)

plt.title(
    "Mental Health Label Distribution"
)

plt.xlabel("Mental Condition")

plt.ylabel("Count")

plt.show()
```

```python
df["status"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title(
    "Mental Health Dataset Distribution"
)

plt.ylabel("")

plt.show()
```

```python
df["text_length"] = df["text"].apply(len)
df["text_length"]
```

```python
df["text_length"].describe()
```

```python
plt.hist(
    df["text_length"],
    bins=100

)

plt.title(
    "Text Length Distribution"
)

plt.xlabel("Length")
plt.ylabel("Frequency")

plt.show()
```
